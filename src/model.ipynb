# Testing different models

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.svm import SVC
from sklearn.ensemble import GradientBoostingClassifier

sns.set(style="whitegrid")

import warnings
warnings.filterwarnings("ignore")
```

Loading data

```python
X_train_processed = pd.read_csv('../data/X_train.csv')
X_test_processed = pd.read_csv('../data/X_test.csv')
X_valid_processed = pd.read_csv('../data/X_valid.csv')
y_train = pd.read_csv('../data/y_train.csv')
y_test = pd.read_csv('../data/y_test.csv')
y_valid = pd.read_csv('../data/y_valid.csv')
```

```python
results = []
```

## Linear Regression

```python
model_lr = LinearRegression(positive=False)

model_lr.fit(X_train_processed, y_train)
y_pred = model_lr.predict(X_valid_processed)
mse_train = mean_squared_error(y_train, model_lr.predict(X_train_processed))
r2_train = r2_score(y_train, model_lr.predict(X_train_processed))
mse_test = mean_squared_error(y_valid, y_pred)
r2_test = r2_score(y_valid, y_pred)

print(f"Linear Regression - Mean Squared Error, Train: {mse_train:.2f}, Test: {mse_test:.2f}")
print(f"Linear Regression - R^2 Score, Train: {r2_train:.2f}, Test: {r2_test:.2f}")
```

## Logistic Regression

```python
model_logr = LogisticRegression()

model_logr.fit(X_train_processed, y_train)
y_pred = model_logr.predict(X_valid_processed)
mse_train = mean_squared_error(y_train, model_logr.predict(X_train_processed))
r2_train = r2_score(y_train, model_logr.predict(X_train_processed))
mse_test = mean_squared_error(y_valid, y_pred)
r2_test = r2_score(y_valid, y_pred)

print("Logistic Regression")
print(f"Logistic Regression - Mean Squared Error, Train: {mse_train:.2f}, Test: {mse_test:.2f}")
print(f"Logistic Regression - R^2 Score, Train: {r2_train:.2f}, Test: {r2_test:.2f}")
print(classification_report(y_valid, model_logr.predict(X_valid_processed)))
print(f"Accuracy score for x_train: {accuracy_score(y_train, model_logr.predict(X_train_processed)):.4f}")
acc_valid = accuracy_score(y_valid, model_logr.predict(X_valid_processed))
print(f"Accuracy score for x_valid: {acc_valid:.4f}")
results.append({
    "model": model_logr.__class__.__name__,
    "accuracy_score": acc_valid
})
```

## RandomForest

```python
rf_model = RandomForestClassifier(
    random_state=40,
    n_estimators=300,
    max_depth=None,
    max_features="sqrt",
    min_samples_split=2,
    min_samples_leaf=1,
    n_jobs=-1
)

rf_model.fit(X_train_processed, y_train)
print("Random Forest")
print(classification_report(y_valid, rf_model.predict(X_valid_processed)))
print(f"Accuracy score for x_train: {accuracy_score(y_train, rf_model.predict(X_train_processed)):.4f}")
acc_valid = accuracy_score(y_valid, rf_model.predict(X_valid_processed))
print(f"Accuracy score for x_valid: {acc_valid:.4f}")
results.append({
    "model": rf_model.__class__.__name__,
    "accuracy_score": acc_valid
})
```

## SVC

```python
svc_model = SVC(C=10, gamma=0.1, kernel='rbf')
svc_model.fit(X_train_processed, y_train)

print("SVC")
print(classification_report(y_valid, svc_model.predict(X_valid_processed)))
print(f"Accuracy score for x_train: {accuracy_score(y_train, svc_model.predict(X_train_processed)):.4f}")
acc_valid = accuracy_score(y_valid, svc_model.predict(X_valid_processed))
print(f"Accuracy score for x_valid: {acc_valid:.4f}")
results.append({
    "model": svc_model.__class__.__name__,
    "accuracy_score": acc_valid
})
```

## Gradient Boosting Classifier

```python
gb_model = GradientBoostingClassifier(random_state=42)
gb_model.fit(X_train_processed, y_train)

print("Gradient Boosting Classifier")
print(classification_report(y_valid, gb_model.predict(X_valid_processed)))
print(f"Accuracy score for x_train: {accuracy_score(y_train, gb_model.predict(X_train_processed)):.4f}")
acc_valid = accuracy_score(y_valid, gb_model.predict(X_valid_processed))
print(f"Accuracy score for x_valid: {acc_valid:.4f}")
results.append({
    "model": gb_model.__class__.__name__,
    "accuracy_score": acc_valid
})
```

## Comparing the results
